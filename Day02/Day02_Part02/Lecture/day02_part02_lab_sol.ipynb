```python
with open('lab04.csv', 'w') as f: # open new csv file
    w = csv.DictWriter(
        f,
        fieldnames=("name", "title", "email", "website", "specialization")
    ) # set colnames
    w.writeheader() # write header

    # set up selenium
    driver = webdriver.Chrome()

    # open webpage
    driver.get('https://polisci.washu.edu/people?cat=88&letter=All') # go to website
    driver.execute_script(
        "window.scrollTo(0, document.body.scrollHeight);"
    ) # scroll to bottom

    time.sleep(5) # pause to let selenium catch up and actually scroll

    html = driver.page_source # get html
    driver.close() # close selenium browser

    soup = BeautifulSoup(html, 'html.parser') # soup html

    cards = soup.find_all('a', {'class': 'card'}) # get all faculty

    for c in range(len(cards)): # for each faculty card
        print("Working on " + str(c) + " of " + str(len(cards)) + ":")

        try:
            fac = {} # empty dict

            fac['name'] = ' '.join(
                cards[c].find('h3').text.split('\xa0')
            ) # get name

            fac['title'] = cards[c].find(
                'div',
                {'class': 'dept'}
            ).text # get title

            interior = (
                'https://polisci.wustl.edu' + cards[c]['href']
            ) # go to interior page

            interior_page = urllib.request.urlopen(interior) # open interior
            interior_soup = BeautifulSoup(
                interior_page.read(),
                'html.parser'
            ) # soup interior

            fac['email'] = interior_soup.find(
                'ul',
                {'class': 'detail contact'}
            ).find('a').text # get email

            fac['website'] = interior_soup.find(
                'ul',
                {'class': 'links'}
            ).find('a')['href'] # get personal website

            fac['specialization'] = interior_soup.find(
                'div',
                {'class': 'post-excerpt'}
            ).text # get specialization

            w.writerow(fac) # write row

        except:
            continue # skip row on issue

        time.sleep(3) # polite sleep

print("All done!")
```

```python
with open('lab04.csv', 'w') as f: # open new csv file
    w = csv.DictWriter(
        f,
        fieldnames=("name", "title", "email", "website", "specialization")
    ) # set colnames
    w.writeheader() # write header

    # set up selenium
    driver = webdriver.Chrome()

    # open ALL people webpage
    driver.get(
        'https://polisci.washu.edu/people?cat=88&letter=All'
    )

    time.sleep(3)

    # keep clicking "Load more" until there is no button
    while True:
        try:
            load_more = driver.find_element(
                By.XPATH,
                "//*[self::a or self::button]"
                "[contains(normalize-space(.), 'Load more')]"
            )

            print("Found Load more!")

            # scroll to button
            driver.execute_script(
                "arguments[0].scrollIntoView();",
                load_more
            )

            time.sleep(1)

            # click Load more
            driver.execute_script(
                "arguments[0].click();",
                load_more
            )

            time.sleep(3)

        except NoSuchElementException:
            print("No more Load more button.")
            break


    # NOW get the final page after all people are loaded
    html = driver.page_source
    driver.close()

    soup = BeautifulSoup(html, 'html.parser')

    cards = soup.find_all(
        'a',
        {'class': 'card'}
    )

    print("Cards before removing duplicates:", len(cards))


    # remove duplicate people using profile URL
    unique_cards = []
    seen = set()

    for card in cards:
        href = card.get('href')

        if href not in seen:
            seen.add(href)
            unique_cards.append(card)

    cards = unique_cards

    print("Total unique people found:", len(cards))


    for c in range(len(cards)): # for each person
        print(
            "Working on "
            + str(c)
            + " of "
            + str(len(cards))
            + ":"
        )

        try:
            fac = {} # empty dict

            fac['name'] = ' '.join(
                cards[c].find('h3').text.split('\xa0')
            ) # get name

            fac['title'] = cards[c].find(
                'div',
                {'class': 'dept'}
            ).text # get title

            interior = (
                'https://polisci.washu.edu'
                + cards[c]['href']
            ) # go to interior page

            interior_page = urllib.request.urlopen(
                interior
            ) # open interior

            interior_soup = BeautifulSoup(
                interior_page.read(),
                'html.parser'
            ) # soup interior


            # get email
            try:
                fac['email'] = interior_soup.find(
                    'ul',
                    {'class': 'detail contact'}
                ).find('a').text
            except:
                fac['email'] = ''


            # get personal website
            try:
                fac['website'] = interior_soup.find(
                    'ul',
                    {'class': 'links'}
                ).find('a')['href']
            except:
                fac['website'] = ''


            # get specialization
            try:
                fac['specialization'] = interior_soup.find(
                    'div',
                    {'class': 'post-excerpt'}
                ).text
            except:
                fac['specialization'] = ''


            w.writerow(fac) # write row


        except Exception as e:
            print("FAILED:", e)
            continue


        time.sleep(3) # polite sleep


print("All done!")
```
